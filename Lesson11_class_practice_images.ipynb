This Python script loads an image, converts it into a numerical grid, isolates the red color channel, and displays the result. Because the green and blue data are removed, the final image appears entirely in shades of red and black.

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
from PIL import Image
```

```python
pic = Image.open('parrot_image.jpg') #Read image file
```

```python
pic
```

```python
pic_arr = np.asarray(pic)
```

```python
pic_arr.shape
```

```python
pic_red = pic_arr.copy()
```

pic_red[:,:,1]=0: NumPy uses 0-based indexing. In the third dimension (the color channels), index 0 is Red, index 1 is Green, and index 2 is Blue. 
This line selects every single pixel ([:,:]) and sets its Green channel (1) to zero.

pic_red[:,:,2]=0: Similarly, this selects every pixel and sets its Blue channel (2) to zero.

Result: Because Green and Blue are entirely erased, only the Red channel numbers remain.

```python
pic_red[:,:,1]=0 # Zero out contribution from green
pic_red[:,:,2]=0 # Zero out contribution from blue
```

```python
pic_red
```

```python
plt.imshow(pic_red)
plt.axis('off') # Hides the pixel coordinate axes
plt.show()
```

**Loading and Displaying an Image with OpenCV**

```python
import cv2 #pip install opencv-python
```

```python
#Load an image
image = cv2.imread('parrot_image.jpg')
```

```python
#Display an image
cv2.imshow('Parrot', image)
cv2.waitKey(0) #wait for any key press
cv2.destroyAllWindows() #Close the window
```

```python
#Splitting into channels
B, G, R = cv2.split(image)
```

```python
# Display the Blue channel
cv2.imshow('Blue Channel', B)
cv2.waitKey(0) #wait for any key press
cv2.destroyAllWindows() #Close the window
```

```python
#Merge channels
merged_image = cv2.merge([B, G, R])
```

```python
# Display the merged image
cv2.imshow('Merged Image', merged_image)
cv2.waitKey(0) #wait for any key press
cv2.destroyAllWindows() #Close the window
```

```python
# Convert to grayscale
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cv2.imshow('Grayscale Image', gray_image)
cv2.waitKey(0) #wait for any key press
cv2.destroyAllWindows() #Close the window
```
